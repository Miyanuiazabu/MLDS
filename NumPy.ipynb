# NumPy Basics

This section introduces some basics of NumPy, a linear algebra and matrix manipulation library.

We begin by importing NumPy, which by convention is imported as `np` to save typing.

```python
import numpy as np
```

## 1-Dimensional Data Structures

First, we will look at 1-dimensional data structures.

```python
v = np.arange(10)
v
```

This represents a 1-dimensional NumPy array (vector) of length 10.

You can access elements using 0-based indexing:

```python
v[3]
```

Accessing a range of values can be done using the `:` operator:

```python
v[0:3]
```

For brevity, you can omit the first index. Omitting the second index means until the last element:

```python
v[:3]
```

Using negative indexing selects elements relative to the end:

```python
v[:-3]
```

```python
v[-3:]
```

Leaving indexes out entirely selects the whole array:

```python
v[:]
```

Array broadcasting applies operations to all elements without loops:

```python
v += 1
v
```

## 2D Arrays (Matrices)

NumPy can handle multidimensional matrices. Indexing is done by rows and columns using a comma.

```python
m = np.arange(9).reshape(3, 3)
```

```python
m
```

### Indexing 2D Arrays

```python
m[0]
```

```python
m[0, 1]
```

```python
m[:, 0]
```

### Advanced Slicing

```python
m[:, ::2]
```

```python
m[::2, ::2]
```

## Element-wise Operations

```python
m**2
```

```python
m + 1
```

### Operations Between 2D and 1D Arrays

```python
n = np.array([1, 2, 3])
n
```

```python
m + n
```

```python
m * n
```
